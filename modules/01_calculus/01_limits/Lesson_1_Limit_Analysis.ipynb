```python
import numpy as np
import matplotlib.pyplot as plt
import os

def run_limit_analysis():
    # Fonksiyon tanımı: x=1'de tanımsız
    x = np.linspace(0.5, 1.5, 100)
    y = (x**2 - 1) / (x - 1)

    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label='f(x) = (x²-1)/(x-1)', color='blue', lw=2)
    plt.scatter([1], [2], color='red', s=100, facecolors='none', label='Hole at L=2')

    # Epsilon-Delta Alanları
    plt.axhspan(1.9, 2.1, alpha=0.1, color='green', label='Epsilon (Error Margin)')
    plt.axvspan(0.95, 1.05, alpha=0.1, color='orange', label='Delta (Input Range)')

    plt.title("Lesson 01: Epsilon-Delta Visualization", fontsize=14)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    
    # Assets klasörüne kaydet
    if not os.path.exists('../../../assets'):
        os.makedirs('../../../assets')
    plt.savefig('../../../assets/01_limit_epsilon.png')
    plt.show()

if __name__ == "__main__":
    run_limit_analysis()
```
